```python

import json
import urllib.request

url = "https://api.coingecko.com/api/v3/coins/bitcoin/market_chart?vs_currency=usd&days=1"

req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
response = urllib.request.urlopen(req, timeout=20)
data = json.loads(response.read().decode("utf-8"))
response.close()

prices = [item[1] for item in data["prices"]]

first_price = prices[0]
last_price = prices[-1]
change = ((last_price - first_price) / first_price) * 100

print("First price:", round(first_price, 2))
print("Latest price:", round(last_price, 2))
print("Price change:", round(change, 2), "%")

trend = "BULLISH" if change > 0.5 else "BEARISH" if change < -0.5 else "SIDEWAYS"

print("Trend:", trend)
```

```python

recent_avg = sum(prices[-6:]) / 6
previous_avg = sum(prices[-12:-6]) / 6

momentum = ((recent_avg - previous_avg) / previous_avg) * 100

print("Recent average:", round(recent_avg, 2))
print("Previous average:", round(previous_avg, 2))
print("Momentum:", round(momentum, 3), "%")

result = "POSITIVE" if momentum > 0 else "NEGATIVE" if momentum < 0 else "NEUTRAL"

print("Second analyst:", result)
```

```python

analyst_1 = "BULLISH" if change > 0.5 else "BEARISH" if change < -0.5 else "SIDEWAYS"; analyst_2 = "BULLISH" if momentum > 0.05 else "BEARISH" if momentum < -0.05 else "SIDEWAYS"; signal = "LONG" if analyst_1 == "BULLISH" and analyst_2 == "BULLISH" else "SHORT" if analyst_1 == "BEARISH" and analyst_2 == "BEARISH" else "NO TRADE"; print("Analyst 1:", analyst_1); print("Analyst 2:", analyst_2); print("FINAL CONSENSUS:", signal)
```

```python

import json
import urllib.request
import pandas as pd

url = "https://api.coinpaprika.com/v1/tickers?quotes=USD"
req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
response = urllib.request.urlopen(req, timeout=30)
market = json.loads(response.read().decode("utf-8"))
response.close()

df = pd.json_normalize(market)
df = df[["rank", "name", "symbol", "quotes.USD.price", "quotes.USD.percent_change_24h", "quotes.USD.volume_24h"]]
df.columns = ["Rank", "Coin", "Symbol", "Price USD", "Change 24h %", "Volume 24h"]
df = df.dropna(subset=["Rank", "Price USD"])
df["Rank"] = pd.to_numeric(df["Rank"])
df = df.sort_values("Rank").head(150).reset_index(drop=True)

print("Number of coins scanned:", len(df))
display(df)
```

```python

stablecoins = ["USDT", "USDC", "BUSD", "FDUSD", "TUSD", "DAI", "USDE", "USDS", "USDD", "PYUSD", "USDP", "EURC"]

df["Symbol"] = df["Symbol"].str.upper()
df["Volume 24h"] = pd.to_numeric(df["Volume 24h"], errors="coerce")
df["Change 24h %"] = pd.to_numeric(df["Change 24h %"], errors="coerce")

candidates = df[~df["Symbol"].isin(stablecoins)].copy()
candidates = candidates.drop_duplicates(subset=["Symbol"])
candidates = candidates.dropna(subset=["Volume 24h", "Price USD"])
candidates = candidates.sort_values("Volume 24h", ascending=False)
top30 = candidates.head(30).reset_index(drop=True)

print("Candidates selected:", len(top30))
display(top30[["Rank", "Coin", "Symbol", "Price USD", "Change 24h %", "Volume 24h"]])
```

```python
excluded_symbols = ["USDT","USDC","BUSD","FDUSD","TUSD","DAI","USDE","USDS","USDD","PYUSD","USDP","EURC","USD1","USDG","UST","USTC","WETH","WBTC","CBBTC","WBNB","BTCB","STETH","WSTETH","RETH","CBETH","WBETH","BETH","XAUT","PAXG"]
```

```python
clean_candidates = candidates[~candidates["Symbol"].isin(excluded_symbols)].copy()
top30_clean = clean_candidates.sort_values("Volume 24h", ascending=False).head(30).reset_index(drop=True)
print("Selected:", len(top30_clean))
display(top30_clean)
```

```python
symbols = top30_clean["Symbol"].tolist()

print("Number of symbols:", len(symbols))
print("Symbols:")
print(", ".join(symbols))
```

```python
code = "def run_portfolio(capital, leverage=3):\n"
code += " equity=float(capital)\n trades=[]\n skipped=0\n last_exit_time=-1\n"
code += " fee=risk_config['fee_rate']\n slip=risk_config['slippage_rate']\n"
code += " for r in best_each.sort_values('entry_time').itertuples():\n"
code += "  s=r.symbol\n  et=int(r.entry_time)\n"
code += "  if et<=last_exit_time: continue\n"
code += "  d=bt15[s].reset_index(drop=True)\n  k=idx_map[s].get(et)\n"
code += "  if k is None or k+96>=len(d):\n   skipped+=1\n   continue\n"
code += "  meta=inst_map[s+'-USDT-SWAP']\n"
code += "  ct=float(meta['ctVal'])\n  lot=float(meta['lotSz'])\n  minsz=float(meta['minSz'])\n"
code += "  minqty=math.ceil(minsz/lot-1e-12)*lot\n"
code += "  raw=float(d.open.iloc[k])\n  side=r.side\n"
code += "  entry=raw*(1-slip) if side=='LONG' else raw*(1+slip)\n"
code += "  stop=float(r.stop)\n  target=float(r.target)\n"
code += "  stop_fill=stop*(1-slip) if side=='LONG' else stop*(1+slip)\n"
code += "  unit_risk=ct*(abs(entry-stop_fill)+fee*(entry+stop_fill))\n"
code += "  if unit_risk<=0:\n   skipped+=1\n   continue\n"
code += "  qty_risk=equity*risk_config['risk_per_trade']/unit_risk\n"
code += "  qty_margin=equity*leverage*0.95/(ct*entry)\n"
code += "  qty=math.floor(min(qty_risk,qty_margin)/lot+1e-10)*lot\n"
code += "  if qty+1e-10<minqty:\n   skipped+=1\n   continue\n"
code += "  exit_raw=None\n  reason='TIME'\n  exit_i=k+96\n"
code += "  for j in range(k,k+97):\n"
code += "   bar=d.iloc[j]\n   op=float(bar.open)\n"
code += "   sh=float(bar.high)>=stop if side=='SHORT' else float(bar.low)<=stop\n"
code += "   th=float(bar.low)<=target if side=='SHORT' else float(bar.high)>=target\n"
code += "   if sh:\n"
code += "    exit_raw=max(stop,op) if side=='SHORT' else min(stop,op)\n"
code += "    reason='SL'\n    exit_i=j\n    break\n"
code += "   if th:\n"
code += "    exit_raw=target\n    reason='TP'\n    exit_i=j\n    break\n"
code += "  if exit_raw is None: exit_raw=float(d.close.iloc[exit_i])\n"
code += "  exit_fill=exit_raw*(1+slip) if side=='SHORT' else exit_raw*(1-slip)\n"
code += "  gross=ct*qty*(entry-exit_fill if side=='SHORT' else exit_fill-entry)\n"
code += "  costs=fee*ct*qty*(entry+exit_fill)\n"
code += "  net=gross-costs\n  equity+=net\n"
code += "  last_exit_time=int(d.time.iloc[exit_i])+900000\n"
code += "  trades.append({'symbol':s,'side':side,'score':float(r.quality_score),'reason':reason,'contracts':qty,'notional':ct*qty*entry,'planned_risk':unit_risk*qty,'net_usdt':net,'equity_after':equity})\n"
code += " return pd.DataFrame(trades),skipped\n"

exec(code)
print("موتور اصلی آماده:",callable(run_portfolio))
```
